Tuple

--> Immutable datatype

--> declare tuple using this ( )

```python
students=('rachana','pujitha','chinni','prahllada','ramadevi')
print(students)
```

```python
type(students)
```

```python
print(students[1])
```

```python
for a in students:
    print(a)
```

```python
dimensions=(10,20)
print(dimensions)
print(type(dimensions))
```

Dictionary

--> combination of key-value pairs

--> mutable datatype

--> declare using { }

```python
alien={'color': 'green','points':5,'level':'stage1'}
print(alien)
```

```python
print(alien['color'])
```

```python
useraccount={'username':'coding','firstname':'lingutla','lastname':'rachana','dob':'24-8-1987','pwd':'532'}
print(useraccount)
```

```python
print(useraccount['username'])
```

```python
useraccount['pwd']='123'
```

```python
print(useraccount)
```

```python
useraccount['firstname']='chinni'
```

```python
print(useraccount)
```
