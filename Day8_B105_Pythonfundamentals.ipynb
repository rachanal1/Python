```python
useraccount={'username':'coding','firstname':'lingutla','lastname':'rachana','dob':'24-8-1987','pwd':'532'}
print(useraccount)
```

```python
for k,v in useraccount.items():
    print(f"key: {k}")
    print(f"vavlue: {v}\n")
```

```python
#only keys
for k in useraccount.keys():
    print(f"key: {k}")
```

```python
#only values
for v in useraccount.values():
    print(f"value: {v}")
```
